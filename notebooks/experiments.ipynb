# MedSight AI — Multimodal Medical Second Opinion Experiments
This notebook demonstrates:
1. Loading a chest X-ray
2. Generating candidate findings via VLM
3. Computing visual grounding heatmaps and bounding boxes
4. Passing candidates through the **Hard Hallucination Gate**
5. Calibrating confidence scores

```python
import sys, os
sys.path.append('..')
import numpy as np
import cv2
import matplotlib.pyplot as plt
from src.reasoning.multimodal_reasoner import MultimodalReasoner

# Initialize Reasoner
reasoner = MultimodalReasoner()
print('Reasoner initialized successfully.')
```

```python
# Create synthetic chest radiograph for demonstration
img = np.zeros((512, 512, 3), dtype=np.uint8)
cv2.ellipse(img, (180, 240), (70, 150), 0, 0, 360, (50, 50, 50), -1)
cv2.ellipse(img, (332, 240), (70, 150), 0, 0, 360, (50, 50, 50), -1)
cv2.ellipse(img, (270, 310), (85, 70), 20, 0, 360, (130, 130, 130), -1)

clinical_notes = 'Patient with hypertension reports dyspnea and orthopnea. History of congestive heart failure.'
results = reasoner.analyze_case(image=img, clinical_notes=clinical_notes)

print(f"Supported findings: {len(results['findings'])}")
print(f"Suppressed hallucinations: {len(results['rejected_findings'])}")
```
